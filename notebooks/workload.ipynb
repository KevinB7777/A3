## Predicted queue wait (one GPU)

```python
WAIT_TARGET_S = 2.0   # P99 wait target used for the rest of the assignment

w = predict_queue_wait(df)
print(f"predicted wait P50 {np.median(w):.2f}s  P99 {np.quantile(w, 0.99):.2f}s  max {w.max():.1f}s")

def longest_late_stretch(t, wait, merge_gap_s=60.0):
    """Longest time range where requests wait over the target (gaps under a minute merged).
    Part 2's notebook uses the same rule."""
    late = t[wait > WAIT_TARGET_S]
    if len(late) == 0:
        return "never"
    breaks = np.flatnonzero(np.diff(late) > merge_gap_s)
    starts, ends = np.r_[late[0], late[breaks + 1]], np.r_[late[breaks], late[-1]]
    i = np.argmax(ends - starts)
    return f"{starts[i]:.0f}-{ends[i]:.0f} s"

print(f"longest stretch above the 2 s target: {longest_late_stretch(df.arrived_at.to_numpy(), w)}")

fig, ax = plt.subplots(figsize=(10, 3.5))
ax.plot(df.arrived_at, w)
ax.axhline(WAIT_TARGET_S, color="k", linestyle="--", linewidth=0.8, label="2 s target")
ax.set_xlabel("trace time (s)")
ax.set_ylabel("predicted wait (s)")
ax.legend(frameon=False)
plt.show()
```

## KV-cache memory in use (one GPU)

```python
print(f"KV per token: {kv_bytes_per_token() / 1024:.0f} KiB   model weights: {wl.MODEL_WEIGHTS_GIB} GiB (transformer layers only)")

kv = kv_in_use(df, w, BIN_S)
print(f"KV in use (estimate): mean {kv.kv_gib.mean():.1f} GiB   peak {kv.kv_gib.max():.1f} GiB "
      f"at t={kv.bin_start[kv.kv_gib.idxmax()]:.0f}s")

fig, ax = plt.subplots(figsize=(10, 3.5))
ax.plot(kv.bin_start, kv.kv_gib)
ax.set_xlabel("trace time (s)")
ax.set_ylabel("estimated KV cache in use (GiB)")
plt.show()
```

Setting `max_batch_size` to the very large value allow us to reach the **peak of 94.4 GiB (at t = 1950 seconds)** which has no GPU with available memory. The peak from above is limited (maximum of 26.7 GiB) which is a more accurate estimate of GPU memory available to tackle requests since the GPU works with a maximum of **128 requests** at once and the request that is already queued does not take anything from the KV cache on the GPU. Calculating with the uncapped value we find the peak that only adds to its memory every time the request is in the queue.

```python
peak_i = kv.kv_gib.idxmax()
print(f"capped (128): mean {kv.kv_gib.mean():.1f} GiB  peak {kv.kv_gib.max():.1f} GiB at t={kv.bin_start[peak_i]:.0f}s")
print(f"{'GPU':22s}{'usable for KV':>15s}{'fits peak':>11s}{'fits average':>14s}")

gpu_list = [("L4 (24 GiB)", 24), ("A100 (40 GiB)", 40), ("L40S (48 GiB)", 48), ("A100 / H100 (80 GiB)", 80)]

for name, gib in gpu_list:
    usable = 0.9 * gib - wl.MODEL_WEIGHTS_GIB
    if kv.kv_gib.max() <= usable:
        p = "yes"
    else:
        p = "no"
        
    if kv.kv_gib.mean() <= usable:
        a = "yes"
    else:
        a = "no"
        
    print(f"{name:22s}{usable:12.1f} GiB{p:>11s}{a:>14s}")

kv_big = kv_in_use(df, w, BIN_S, max_batch_size=10**9)

t_peak = kv_big.bin_start[kv_big.kv_gib.idxmax()]
max_run = kv_big.running.max()

print(f"uncapped: peak {kv_big.kv_gib.max():.1f} GiB at t={t_peak:.0f}s  (most requests in the system at once: {max_run})")
```

**Q3.** At 128 tokens per 30 ms step, one GPU decodes at most 4,267 tokens/s. What is that throughput if a step takes 20 ms or 50 ms, or if the batch holds only 64? Rerun `predict_queue_wait` with each. How much does the P99 wait move compared with the change in throughput, and why? Then set `decode_tok_per_s=float("inf")`: in this model, can faster decode alone meet the 2 s target? If not, what is the bottleneck?

**Your answer:**

Baseline 30 ms step, batch 128: 4,267 tokens/s and P99 wait 72.1 s.

| Decode setting | Throughput | Change | P99 wait | Change |
|---|---|---|---|---|
| step 20 ms, batch 128 | 6,400 tokens/s | +50% | 30.7 s | -57% |
| step 50 ms, batch 128 | 2,560 tokens/s | -40% | 201.4 s | +179% |
| step 30 ms, batch 64 | 2,133 tokens/s | -50% | 386.4 s | +436% |

The P99 wait time is quite sensitive in comparison to throughput. A halving of the throughput makes the P99 wait time five times worse. The trace is nearly at the GPU's capacity; in the best possible case the trace requires 1.01 seconds of GPU time to process one second's worth of work. After the load exceeds capacity, time spent waiting is calculated as the backlog of time built during the surge (which is effectively (load - capacity) * ramification). The small change translates into a huge backlog. 

With decode_tok_per_s=float("inf"), the value for P99 wait time is still 3.4 seconds (max 4.9 seconds with 4.3% of the requests exceeding 2 seconds); therefore the efforts to speed up decoding won't suffice to achieve the 2 seconds criterion. Prefill is the main bottleneck here - out of the overall processing time of 2799 seconds prompts take 1840 seconds while the rate of incoming prompts can reach 13000 tokens per second in case of surge, this means that the occupancy of the queue will never go away regardless of the decoding speed.


```python
# Q3 scratch: predict_queue_wait(df, decode_tok_per_s=batch / step_s)
t = df.arrived_at.to_numpy()

base_wait = predict_queue_wait(df)
base = np.quantile(base_wait, 0.99)

print(f"{'setting':24s}{'tokens/s':>10s}{'change':>8s}{'P99 wait':>10s}{'change':>8s}   stretch above 2 s")

scenarios = [
    ("baseline (30 ms, 128)", 128, 30e-3), 
    ("step 20 ms, batch 128", 128, 20e-3),
    ("step 50 ms, batch 128", 128, 50e-3), 
    ("step 30 ms, batch 64", 64, 30e-3)
]

for label, batch_size, step_time in scenarios:
    tp = batch_size / step_time
    ww = predict_queue_wait(df, decode_tok_per_s=tp)
    p99 = np.quantile(ww, 0.99)
    
    tp_change = tp / wl.DECODE_TOK_PER_S - 1
    p99_change = p99 / base - 1
    stretch_result = longest_late_stretch(t, ww)
    
    print(f"{label:24s}{tp:10,.0f}{tp_change:+8.0%}{p99:9.1f}s{p99_change:+8.0%}   {stretch_result}")

winf = predict_queue_wait(df, decode_tok_per_s=float("inf"))
winf_p99 = np.quantile(winf, 0.99)
winf_max = winf.max()
winf_over_target = np.mean(winf > WAIT_TARGET_S)
winf_stretch = longest_late_stretch(t, winf)

print(f"decode_tok_per_s=inf: P99 {winf_p99:.1f}s  max {winf_max:.1f}s  "
      f"over 2 s {winf_over_target:.1%}  stretch {winf_stretch}")

prompt_work = (df.num_prefill_tokens / wl.PREFILL_TOK_PER_S).sum()
total_work = prompt_work + (df.num_decode_tokens / wl.DECODE_TOK_PER_S).sum()

print(f"work: prompts {prompt_work:,.0f} s of {total_work:,.0f} s total; trace is {span:,.0f} s long")
```
